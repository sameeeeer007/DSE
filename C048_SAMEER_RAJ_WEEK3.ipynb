# EXPERIMENT-1

```python
from collections import deque

def bfs(graph, start, goal):

    queue = deque()
    queue.append([start])

  
    visited = set()


    nodes_expanded = 0

    while queue:
     
        path = queue.popleft()
        current_node = path[-1]


        if current_node == goal:
            return path, nodes_expanded

   
        if current_node not in visited:
            visited.add(current_node)
            nodes_expanded += 1

            for neighbor in graph[current_node]:
                new_path = path + [neighbor]
                queue.append(new_path)

  
    return None, nodes_expanded


maze = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E', 'G'],
    'G': []
}

start_node = 'A'
goal_node = 'G'

path, expanded = bfs(maze, start_node, goal_node)

print("Path found:", path)
print("Nodes expanded:", expanded)
```

# EXPERIMENT-2

```python
def dfs(graph, current, goal, path):

    path.append(current)


    if current == goal:
        return path

    for neighbor in graph[current]:
    
        if neighbor not in path:
            result = dfs(graph, neighbor, goal, path)
            if result is not None:
                return result


    path.pop()
    return None



graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E', 'G'],
    'G': []
}


start_node = 'A'
goal_node = 'G'


path = dfs(graph, start_node, goal_node, [])


if path:
    print("Path found using DFS:", path)
else:
    print("No path found")
```

# EXPERIMENT-3

```python
from collections import deque


def get_neighbors(state):
    neighbors = []
    index = state.index(0)  
    row = index // 3
    col = index % 3

    moves = {
        'UP': (-1, 0),
        'DOWN': (1, 0),
        'LEFT': (0, -1),
        'RIGHT': (0, 1)
    }

    for move in moves.values():
        new_row = row + move[0]
        new_col = col + move[1]

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_index = new_row * 3 + new_col
            new_state = list(state)
           
            new_state[index], new_state[new_index] = new_state[new_index], new_state[index]
            neighbors.append(tuple(new_state))

    return neighbors



def bfs(initial_state, goal_state):
    queue = deque()
    queue.append([initial_state])

    visited = set()
    visited.add(initial_state)

    states_expanded = 0

    while queue:
        path = queue.popleft()
        current_state = path[-1]
        states_expanded += 1

        
        if current_state == goal_state:
            return path, states_expanded

       
        for neighbor in get_neighbors(current_state):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(path + [neighbor])

    return None, states_expanded



def print_state(state):
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()



initial_state = (
    1, 2, 3,
    4, 0, 5,
    6, 7, 8
)

goal_state = (
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
)


solution_path, expanded = bfs(initial_state, goal_state)


if solution_path:
    print("Solution Path:\n")
    for state in solution_path:
        print_state(state)

    print("Total states expanded:", expanded)
    print("Number of moves:", len(solution_path) - 1)
else:
    print("No solution found")
```

# EXPERIMENT-4

```python
from collections import deque
import time


graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E', 'G'],
    'G': []
}


def bfs(graph, start, goal):
    queue = deque([[start]])
    visited = set([start])

    nodes_expanded = 0
    max_queue_size = 1

    while queue:
        max_queue_size = max(max_queue_size, len(queue))
        path = queue.popleft()
        current = path[-1]
        nodes_expanded += 1

        if current == goal:
            return path, nodes_expanded, max_queue_size

        for neighbor in graph[current]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(path + [neighbor])

    return None, nodes_expanded, max_queue_size



def dfs(graph, current, goal, path, visited, depth_info):
    visited.add(current)
    path.append(current)

    depth_info["nodes_expanded"] += 1
    depth_info["max_depth"] = max(depth_info["max_depth"], len(path))

    if current == goal:
        return path

    for neighbor in graph[current]:
        if neighbor not in visited:
            result = dfs(graph, neighbor, goal, path, visited, depth_info)
            if result:
                return result

    path.pop()
    return None


goal_node = 'G'

start_time = time.time()
bfs_path, bfs_nodes, bfs_memory = bfs(graph, start_node, goal_node)
bfs_time = time.time() - start_time


dfs_info = {"nodes_expanded": 0, "max_depth": 0}
start_time = time.time()
dfs_path = dfs(graph, start_node, goal_node, [], set(), dfs_info)
dfs_time = time.time() - start_time


print("\nPERFORMANCE COMPARISON\n")

print("BFS Results:")
print("Path:", bfs_path)
print("Nodes Expanded:", bfs_nodes)
print("Execution Time:", bfs_time)
print("Maximum Memory Used (Queue Size):", bfs_memory)

print("\nDFS Results:")
print("Path:", dfs_path)
print("Nodes Expanded:", dfs_info["nodes_expanded"])
print("Execution Time:", dfs_time)
print("Maximum Memory Used (Recursion Depth):", dfs_info["max_depth"])
```

# Post-Exercise Analysis Questions

# 1. Why does BFS guarantee an optimal solution while DFS does not?

Breadth-First Search guarantees an optimal solution because it explores the search space level by level. In an unweighted graph, all actions have the same cost, so the shallowest goal node corresponds to the shortest path. BFS always finds the goal at the minimum depth first, ensuring optimality. Depth-First Search, however, explores one branch completely before considering others and may reach the goal through a longer path even when a shorter path exists elsewhere, so it does not guarantee an optimal solution.

# 2. In which scenarios is DFS preferred over BFS?

Depth-First Search is preferred in scenarios where memory is limited because it stores only the current path rather than all nodes at a given level. It is also useful when the solution is expected to be deep in the search tree and when finding the shortest path is not a requirement. DFS is commonly used in problems involving backtracking, such as maze solving, puzzle solving, and constraint satisfaction problems.

# 3. How does the branching factor affect BFS performance?

The branching factor, which represents the average number of child nodes per state, has a significant impact on BFS performance. As the branching factor increases, the number of nodes generated at each level grows exponentially. This causes BFS to consume a large amount of time and memory because it must store and expand all nodes at each depth before moving to the next level.

# 4. Why is BFS considered impractical for large state spaces?

BFS is considered impractical for large state spaces because it requires a huge amount of memory to store all frontier nodes at each level of the search tree. Its time and space complexity grow exponentially with the depth of the solution and the branching factor. As a result, BFS becomes inefficient and difficult to scale for large or complex problems, even though it guarantees an optimal solution.

```python
# Program to create and print adjacency list

# Take number of vertices and edges
vertices = int(input("Enter number of vertices: "))
edges = int(input("Enter number of edges: "))

# Create empty adjacency list dictionary
graph = {}

# Initialize dictionary with empty lists
for i in range(vertices):
    vertex = input(f"Enter name of vertex {i+1}: ")
    graph[vertex] = []

# Take edges input
print("\nEnter edges (source destination):")

for i in range(edges):
    src, dest = input().split()
    
    # Since graph is undirected, add both sides
    graph[src].append(dest)
    graph[dest].append(src)

# Print adjacency list
print("\nAdjacency List:")
for vertex in graph:
    print(vertex, "->", graph[vertex])
```

```python
# Program to create and print adjacency matrix

# Input number of vertices and edges
vertices = int(input("Enter number of vertices: "))
edges = int(input("Enter number of edges: "))

# Store vertex names
vertex_list = []

for i in range(vertices):
    name = input(f"Enter name of vertex {i+1}: ")
    vertex_list.append(name)

# Create matrix initialized with 0
matrix = [[0 for _ in range(vertices)] for _ in range(vertices)]

print("\nEnter edges (source destination):")

for i in range(edges):
    src, dest = input().split()
    
    # Find index of vertices
    src_index = vertex_list.index(src)
    dest_index = vertex_list.index(dest)
    
    # Since graph is undirected
    matrix[src_index][dest_index] = 1
    matrix[dest_index][src_index] = 1

# Print adjacency matrix
print("\nAdjacency Matrix:\n")

# Print header
print("   ", end="")
for v in vertex_list:
    print(v, end=" ")
print()

for i in range(vertices):
    print(vertex_list[i], end=" ")
    for j in range(vertices):
        print(matrix[i][j], end=" ")
    print()
```
